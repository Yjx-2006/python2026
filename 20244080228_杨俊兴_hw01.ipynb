3、欧几里得算法

3.1 实现gcd函数

```python
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return m
```

3.2 评估三组数字对的 GCD

```python
print(gcd(2018, 2019))  
print(gcd(1200, 300))   
print(gcd(5040, 60))    
```

3.3 参数为负

```python
gcd(-2018, 2019)   # -2018 % 2019 = 1  → 返回 1
gcd(2018, -2019)   # 2018 % -2019 = -1 → 返回 -1
gcd(-1200, -300)   # -1200 % -300 = 0  → 返回 -300
gcd(-1200, 300)    # -1200 % 300 = 0   → 返回 300
```

数学上没有意义，但在程序实现上是合理的（可预测的）。

数学角度：GCD 在数学上定义为正整数，即"能同时整除 m 和 n 的最大正整数"。所以 gcd(-1200, -300) 数学上应该返回 300，而不是 -300。

代码角度：算法只关心"整除关系"，符号并不影响整除性。所以代码能正确找出约数的"绝对值大小"，只是可能带有负号。

4.近似欧拉数e

4.1 euler_limit

```python
def euler_limit(n):
    return (1 + 1 / n) ** n

# 测试
print(euler_limit(1000))
print(euler_limit(1000000))
```

4.2 euler_infinite_sum

```python
def euler_infinite_sum(n):
    total = 0.0
    factorial = 1  # 0! = 1
    for k in range(n):
        if k > 0:
            factorial *= k  # 更新为 k!
        total += 1 / factorial
    return total

# 测试
print(euler_infinite_sum(0))  # 0
print(euler_infinite_sum(1))  # 1.0
print(euler_infinite_sum(4))  # 1 + 1 + 1/2 + 1/6 = 2.6666...
```

4.3 euler_approx

```python
import math

def euler_approx(tolerance):
    total = 0.0
    factorial = 1
    k = 0
    true_e = math.exp(1)
    while abs(total - true_e) > tolerance:
        if k > 0:
            factorial *= k
        total += 1 / factorial
        k += 1
    return total

# 测试
print(euler_approx(0.01))
print(euler_approx(0.0001))
print(euler_approx(1e-10))
```

4.4 print_euler_sum_table

```python
def euler_infinite_sum(n):
    total = 0.0
    factorial = 1
    for k in range(n):
        if k > 0:
            factorial *= k
        total += 1 / factorial
    return total

def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(euler_infinite_sum(k))

# 测试
print_euler_sum_table(10)
```

5.测试整数属性

5.1 is_power_of_2

```python
def is_power_of_2(n):
    while n > 1:
        if n % 2 != 0:      # 除不尽，说明不是 2 的幂
            return False
        n = n // 2          # 整除 2
    return True             # 最终得到 1

# 测试
print(is_power_of_2(1))    # True  (2^0)
print(is_power_of_2(2))    # True
print(is_power_of_2(4))    # True
print(is_power_of_2(8))    # True
print(is_power_of_2(6))    # False
print(is_power_of_2(100))  # False
```

5.2 is_power(b, n)

```python
def is_power(b, n):
    while n > 1:
        if n % b != 0:      # 除不尽，说明 n 不是 b 的幂
            return False
        n = n // b          # 整除 b
    return True             # 最终得到 1

# 测试
print(is_power(2, 8))     # True   (2^3 = 8)
print(is_power(3, 9))     # True   (3^2 = 9)
print(is_power(3, 27))    # True   (3^3 = 27)
print(is_power(5, 125))   # True   (5^3 = 125)
print(is_power(2, 10))    # False
print(is_power(3, 10))    # False
print(is_power(10, 1))    # True   (b^0 = 1)
```
