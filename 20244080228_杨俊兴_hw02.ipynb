1.字符的乐趣

1.1 回⽂是指⼀个词或短语，它从前往后读和从后往前读都⼀样。例如，level、
kayak 和 pop 都是英⽂回⽂；如果忽略空格和标点符号，rats live on no
evil star 和 Was it a car or a cat I saw? 也是回⽂短语。编写⼀个名
为 is_palindrome 的函数，它接受⼀个字符串作为唯⼀参数，并返回布尔值：如
果输⼊是回⽂，返回 True，否则返回 False。您可以假设输⼊仅包含英⽂字⺟
（不区分⼤⼩写）、数字 0 到 9 和空格。判断回⽂时应忽略空格和⼤⼩写，因此
tacocat 和 T A C O cat 都应被认为是回⽂。

```python
def is_palindrome(text):
    normalized = ''.join(text.lower().split())
    return normalized == normalized[::-1]

for s in ["level", "kayak", "pop", "tacocat", "T A C O cat", "rats live on no evil star"]:
    print(repr(s), is_palindrome(s))
```

1.2 如果⼀个词的字⺟从左到右按字⺟顺序排列，就称它是“字⺟顺序的”；允许字⺟重
复。例如，adder 和 beet 是字⺟顺序的，dog 和 cat 则不是。编写⼀个名为
is_abecedarian 的函数，它只接受⼀个字符串参数。如果参数符合上述要求，
返回 True，否则返回 False。您可以假设输⼊仅由字⺟和空格组成。判断时应忽
略空格，因此 abcd efgh xyz 被认为是字⺟顺序的。

```python
def is_abecedarian(text):
    letters = ''.join(text.split()).lower()
    return all(letters[i] <= letters[i+1] for i in range(len(letters)-1))

for s in ["adder", "beet", "dog", "cat", "abcd efgh xyz", ""]:
    print(repr(s), is_abecedarian(s))
```

1.3编写⼀个名为 remove_vowels 的函数，它接受⼀个字符串作为唯⼀参数，并返回
去掉所有元⾳后的字符串。本题的元⾳是字⺟ a、e、i、o、u。例如，
remove_vowels('cat') 应返回 'ct'，remove_vowels('audio') 应返回
'd'。函数也应正确处理全部由元⾳组成的字符串。

```python
def remove_vowels(text):
    vowels = set("aeiouAEIOU")
    return ''.join(ch for ch in text if ch not in vowels)

for s in ["cat", "audio", "AEIOU", "rhythm", "Beautiful"]:
    print(repr(s), "->", repr(remove_vowels(s)))
```

2.列表的乐趣

2.1编写⼀个名为 list_reverse 的函数，它接受⼀个列表作为参数，并返回反转后
的列表。例如，输⼊ [1, 2, 3]，应返回 [3, 2, 1]。如果输⼊不是列表，函数
应引发适当的错误。

```python
def list_reverse(items):
    if not isinstance(items, list):
        raise TypeError("list_reverse 的参数必须是 list")
    return items[::-1]

print(list_reverse([1, 2, 3]))
try:
    list_reverse((1, 2, 3))
except TypeError as e:
    print(type(e).__name__, e)
```

2.2编写⼀个名为 is_sorted 的函数，它接受⼀个序列 seq 作为唯⼀参数。如果序列
按升序排列，返回 True，否则返回 False。您可以假设 seq 是 Python 序列，且
所有元素都⽀持⽐较操作，因此⽆需检查输⼊类型。函数只应遍历序列⼀次；需要
多次遍历的低效解法不能获得满分。这个问题展⽰了 Python 动态类型的特点：只
要 seq ⽀持索引，且元素⽀持⽐较，就⽆需预先知道序列的具体类型。

```python
def is_sorted(seq):
    for i in range(1, len(seq)):
        if seq[i] < seq[i-1]:
            return False
    return True

for s in [[1, 2, 3], [1, 1, 2], [3, 2, 1], "abc", "acb", []]:
    print(repr(s), is_sorted(s))
```

3.作为向量的元组

3.1 实现⼀个名为 vec_scalar_mult 的函数。它接受两个参数：⼀个由整数或浮点
数组成的元组 t，以及⼀个整数或浮点数 s。函数返回与 t ⻓度相同的元组，每个
元素等于 t 的对应元素乘以 s。函数应检查输⼊类型；类型不合适时，引发带有适
当错误消息的 TypeError。整数标量和含有整数元素的元组都是有效输⼊。

```python
import numbers

def _is_numeric_tuple(t):
    return isinstance(t, tuple) and all(
        isinstance(x, numbers.Real) and not isinstance(x, bool) for x in t
    )

def vec_scalar_mult(t, s):
    if not _is_numeric_tuple(t):
        raise TypeError("t 必须是只包含整数或浮点数的元组")
    if not isinstance(s, numbers.Real) or isinstance(s, bool):
        raise TypeError("s 必须是整数或浮点数")
    return tuple(x * s for x in t)

print(vec_scalar_mult((1, 2.5, -3), 2))
print(vec_scalar_mult((2, 4), 0.5))
```

3.2 实现⼀个名为 vec_inner_product 的函数。它接受两个由整数或浮点数组成的
元组，返回这两个向量内积对应的浮点数。向量 的内积是
。若任⼀输⼊不是元组，应引发 TypeError；若两个元组的⻓度不⼀致，应引发
ValueError。

```python
def vec_inner_product(t1, t2):
    if not _is_numeric_tuple(t1) or not _is_numeric_tuple(t2):
        raise TypeError("两个参数都必须是数值元组")
    if len(t1) != len(t2):
        raise ValueError("两个向量的长度必须相同")
    return float(sum(a*b for a, b in zip(t1, t2)))

print(vec_inner_product((1, 2, 3), (4, 5, 6)))
try:
    vec_inner_product((1, 2), (3,))
except ValueError as e:
    print(type(e).__name__, e)
```

3.3 将上述表⽰⽅法扩展到矩阵。⽤⼀个元组的元组表⽰ 矩阵 my_mx：外层元
组⻓度为 ，每个内层元组⻓度为 ，表⽰矩阵的⼀⾏。编写⼀个名为
check_valid_mx 的函数，它接受⼀个参数；若该参数是有效矩阵则返回 True，
否则返回 False。有效矩阵满⾜：
外层元组的每个元素本⾝都是元组；
所有内层元组⻓度相同；
每个内层元组的每个元素都是整数或浮点数。

```python
def check_valid_mx(mx):
    if not isinstance(mx, tuple):
        return False
    if not all(isinstance(row, tuple) for row in mx):
        return False
    if not mx:
        return True
    n = len(mx[0])
    if any(len(row) != n for row in mx):
        return False
    return all(isinstance(x, numbers.Real) and not isinstance(x, bool)
               for row in mx for x in row)

for mx in [((1, 2), (3, 4)), ((1, 2), (3,)), ((1, "x"), (3, 4)), [[1, 2], [3, 4]], ((), ())]:
    print(repr(mx), check_valid_mx(mx))
```

3.4 编写⼀个名为 mx_vec_mult 的函数，它接受⼀个矩阵（元组的元组）和⼀个向量
（元组），返回向量与矩阵相乘得到的数值元组。函数check_valid_mx应检查参
数是否合理；对于不合理的输⼊，应引发适当的错误。

```python
def mx_vec_mult(mx, vec):
    if not check_valid_mx(mx):
        raise TypeError("mx 必须是有效的数值元组矩阵")
    if not _is_numeric_tuple(vec):
        raise TypeError("vec 必须是只包含整数或浮点数的元组")
    n_cols = len(mx[0]) if mx else 0
    if len(vec) != n_cols:
        raise ValueError("向量长度必须等于矩阵列数")
    return tuple(float(sum(a*b for a, b in zip(row, vec))) for row in mx)

print(mx_vec_mult(((1, 2, 3), (4, 5, 6)), (1, 0, -1)))
```

4.更多向量的乐趣

4.1定义⼀个名为 Vector 的类。每个向量都有⼀个维度（⾮负整数）和⼀个元素列表
或元组。初始化函数的第⼀个参数是维度，第⼆个参数是由整数或浮点数组成的列
表或元组，表⽰向量的元素。如果只提供维度⽽不提供元素，请选择合理的默认⾏
为。若维度的类型不正确或为负数，应抛出合适的异常；若维度与元素个数不⼀
致，也应抛出异常。

```python
class Vector:
    def __init__(self, dimension, elements=None):
        if not isinstance(dimension, int) or isinstance(dimension, bool):
            raise TypeError("dimension 必须是非负整数")
        if dimension < 0:
            raise ValueError("dimension 不能为负数")
        if elements is None:
            values = (0,) * dimension
        else:
            if not isinstance(elements, (list, tuple)):
                raise TypeError("elements 必须是 list 或 tuple")
            if not all(isinstance(x, numbers.Real) and not isinstance(x, bool) for x in elements):
                raise TypeError("向量元素必须是整数或浮点数")
            if len(elements) != dimension:
                raise ValueError("dimension 必须与元素个数一致")
            values = tuple(elements)
        self.dimension = dimension
        self.elements = tuple(values)

    def __repr__(self):
        return f"Vector({self.dimension}, {self.elements})"

print(Vector(3))
print(Vector(2, [2, 3]))
```

4.2 维度和元素应是类属性还是实例属性？说明选择的理由。

它们应是实例属性，因为不同向量对象可以具有不同维度和坐标；若作为类属性，会在实例之间共享，造成数据混淆。

4.3 实现向量对象的⽐较运算符，包括相等、⼩于、⼩于或等于、⼤于等。两个
Vector 对象的坐标相同，则视为相等；否则按 Python 元组的字典序⽐较，即先
⽐较第⼀个坐标，再⽐较第⼆个坐标，依此类推。例如，⼆维向量 (2, 4) ⼩于
(2, 5)。⽐较维度不同的两个向量时，应抛出异常。

```python
class Vector:
    def __init__(self, dimension, elements=None):
        if not isinstance(dimension, int) or isinstance(dimension, bool):
            raise TypeError("dimension 必须是非负整数")
        if dimension < 0:
            raise ValueError("dimension 不能为负数")
        if elements is None:
            values = (0,) * dimension
        else:
            if not isinstance(elements, (list, tuple)):
                raise TypeError("elements 必须是 list 或 tuple")
            if not all(isinstance(x, numbers.Real) and not isinstance(x, bool) for x in elements):
                raise TypeError("向量元素必须是整数或浮点数")
            if len(elements) != dimension:
                raise ValueError("dimension 必须与元素个数一致")
            values = tuple(elements)
        self.dimension, self.elements = dimension, tuple(values)

    def _check(self, other):
        if not isinstance(other, Vector):
            raise TypeError("参数必须是 Vector 对象")
        if self.dimension != other.dimension:
            raise ValueError("不能比较维度不同的向量")

    def __eq__(self, other):
        if not isinstance(other, Vector): return NotImplemented
        self._check(other); return self.elements == other.elements
    def __lt__(self, other):
        self._check(other); return self.elements < other.elements
    def __le__(self, other):
        self._check(other); return self.elements <= other.elements
    def __gt__(self, other):
        self._check(other); return self.elements > other.elements
    def __ge__(self, other):
        self._check(other); return self.elements >= other.elements
    def __repr__(self):
        return f"Vector({self.dimension}, {self.elements})"

v1, v2, v3 = Vector(2, (2, 4)), Vector(2, (2, 5)), Vector(2, (2, 4))
print(v1 == v3, v1 < v2, v1 <= v3, v2 > v1, v2 >= v1)
try:
    print(v1 < Vector(3, (1, 2, 3)))
except ValueError as e:
    print(type(e).__name__, e)
```

4.4实现 Vector.dot ⽅法，接收另⼀个 Vector 对象，返回两向量的内积。若参数
类型不正确或维度不⼀致，应抛出合适的异常。

```python
class Vector:
    def __init__(self, dimension, elements=None):
        if not isinstance(dimension, int) or isinstance(dimension, bool):
            raise TypeError("dimension 必须是非负整数")
        if dimension < 0: raise ValueError("dimension 不能为负数")
        if elements is None: values = (0,) * dimension
        else:
            if not isinstance(elements, (list, tuple)): raise TypeError("elements 必须是 list 或 tuple")
            if not all(isinstance(x, numbers.Real) and not isinstance(x, bool) for x in elements):
                raise TypeError("向量元素必须是整数或浮点数")
            if len(elements) != dimension: raise ValueError("dimension 必须与元素个数一致")
            values = tuple(elements)
        self.dimension, self.elements = dimension, tuple(values)
    def _check(self, other):
        if not isinstance(other, Vector): raise TypeError("参数必须是 Vector 对象")
        if self.dimension != other.dimension: raise ValueError("两个向量的维度必须相同")
    def dot(self, other):
        self._check(other)
        return sum(a*b for a, b in zip(self.elements, other.elements))
    def __repr__(self): return f"Vector({self.dimension}, {self.elements})"

print(Vector(3, (1, 2, 3)).dot(Vector(3, (4, 5, 6))))
```

4.5实现 Vector 的乘法运算符。标量左乘或右乘（如 2*v 和 v*2）应返回⼀个各元
素都乘以该标量的新 Vector 对象。当两个 Vector 对象相乘（v*w）时，对应位
置的元素逐⼀相乘，结果仍是⼀个新 Vector 对象；例如，Vector(2, (2, 3))
* Vector(2, (4, 5)) 的元素为 (8, 15)。若两个向量的维度不⼀致，应抛出
合适的异常。

```python
class Vector:
    def __init__(self, dimension, elements=None):
        if not isinstance(dimension, int) or isinstance(dimension, bool):
            raise TypeError("dimension 必须是非负整数")
        if dimension < 0: raise ValueError("dimension 不能为负数")
        if elements is None: values = (0,) * dimension
        else:
            if not isinstance(elements, (list, tuple)): raise TypeError("elements 必须是 list 或 tuple")
            if not all(isinstance(x, numbers.Real) and not isinstance(x, bool) for x in elements):
                raise TypeError("向量元素必须是整数或浮点数")
            if len(elements) != dimension: raise ValueError("dimension 必须与元素个数一致")
            values = tuple(elements)
        self.dimension, self.elements = dimension, tuple(values)
    def _check(self, other):
        if not isinstance(other, Vector): raise TypeError("参数必须是 Vector 对象")
        if self.dimension != other.dimension: raise ValueError("两个向量的维度必须相同")
    def dot(self, other):
        self._check(other)
        return sum(a*b for a, b in zip(self.elements, other.elements))
    def __mul__(self, other):
        if isinstance(other, numbers.Real) and not isinstance(other, bool):
            return Vector(self.dimension, tuple(x*other for x in self.elements))
        if isinstance(other, Vector):
            self._check(other)
            return Vector(self.dimension, tuple(a*b for a, b in zip(self.elements, other.elements)))
        raise TypeError("Vector 只能乘以数值标量或另一个 Vector")
    def __rmul__(self, other):
        if isinstance(other, numbers.Real) and not isinstance(other, bool):
            return Vector(self.dimension, tuple(other*x for x in self.elements))
        return NotImplemented
    def __repr__(self):
        return f"Vector({self.dimension}, {self.elements})"

v, w = Vector(2, (2, 3)), Vector(2, (4, 5))
print("2 * v =", 2*v)
print("v * 2 =", v*2)
print("v * w =", v*w)
print("v.dot(w) =", v.dot(w))
try:
    print(v * Vector(3, (1, 2, 3)))
except ValueError as e:
    print(type(e).__name__, e)
```
