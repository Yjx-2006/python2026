实验2：函数、条件、循环与递归

```python
print("=" * 60)
print("任务1：成绩等级函数")
print("=" * 60)

def score_level(score):
    """根据分数返回等级，超出0~100范围抛出ValueError"""
    if not (0 <= score <= 100):
        raise ValueError(f"分数 {score} 超出有效范围 0~100")
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"

# 测试
test_scores = [95, 85, 70, 55, 101, -1]
for s in test_scores:
    try:
        print(f"  分数 {s:>4} -> {score_level(s)}")
    except ValueError as e:
        print(f"  分数 {s:>4} -> 异常: {e}")

print("\n" + "=" * 60)
print("任务2：for 与 while 求和比较")
print("=" * 60)

# for 循环实现
sum_for = 0
for i in range(1, 101):
    sum_for += i
print(f"  for 循环求和: 1+2+...+100 = {sum_for}")

# while 循环实现
sum_while = 0
j = 1
while j <= 100:
    sum_while += j
    j += 1
print(f"  while 循环求和: 1+2+...+100 = {sum_while}")

print(f"  两种结果是否一致: {sum_for == sum_while}")

print("\n" + "=" * 60)
print("任务3：阶乘函数（递归 vs 迭代）")
print("=" * 60)

def factorial_recursive(n):
    """递归版阶乘，约定 0! = 1"""
    if n < 0 or not isinstance(n, int):
        raise ValueError("输入必须为非负整数")
    if n == 0 or n == 1:
        return 1
    return n * factorial_recursive(n - 1)

def factorial_iterative(n):
    """迭代版阶乘，约定 0! = 1"""
    if n < 0 or not isinstance(n, int):
        raise ValueError("输入必须为非负整数")
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

# 比较 n = 0, 1, 5, 10 时的结果
test_n = [0, 1, 5, 10]
print(f"  {'n':>4} | {'递归版':>12} | {'迭代版':>12} | {'一致':>6}")
print(f"  {'-'*4}-+-{'-'*12}-+-{'-'*12}-+-{'-'*6}")
for n in test_n:
    r_rec = factorial_recursive(n)
    r_itr = factorial_iterative(n)
    match = "✓" if r_rec == r_itr else "✗"
    print(f"  {n:>4} | {r_rec:>12} | {r_itr:>12} | {match:>6}")

print("\n" + "=" * 60)
print("任务4：summary 函数（最小值、最大值、平均值）")
print("=" * 60)

def summary(values):
    """接收非空数值列表，返回 (最小值, 最大值, 平均值) 元组"""
    if not values:
        raise ValueError("列表不能为空")
    min_val = min(values)
    max_val = max(values)
    avg_val = sum(values) / len(values)
    return (min_val, max_val, avg_val)

# 测试并通过元组解包接收结果
data = [12, 45, 7, 23, 89, 34, 56]
min_v, max_v, avg_v = summary(data)
print(f"  数据: {data}")
print(f"  最小值: {min_v}")
print(f"  最大值: {max_v}")
print(f"  平均值: {avg_v:.2f}")

print("\n" + "=" * 60)
print("任务5：sum_all(*args) 可变参数求和")
print("=" * 60)

def sum_all(*args):
    """返回所有参数之和"""
    return sum(args)

# 测试无参数、单个参数、多个参数
print(f"  sum_all()          = {sum_all()}")
print(f"  sum_all(5)         = {sum_all(5)}")
print(f"  sum_all(1, 2, 3)   = {sum_all(1, 2, 3)}")
print(f"  sum_all(10, 20, 30, 40, 50) = {sum_all(10, 20, 30, 40, 50)}")
```

实验3：列表、字典、元组与基础统计

```python
# ============================================================
# 实验3：列表、字典、元组
# 模拟成绩数据
# ============================================================
names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]

# ============================================================
# 任务1：使用 zip() 构建字典，使用 items() 逐项输出
# ============================================================
print("=" * 60)
print("任务1：zip() 构建字典 + items() 遍历")
print("=" * 60)

score_dict = dict(zip(names, scores))
print(f"生成的字典: {score_dict}")
print("\n逐项输出：")
for name, score in score_dict.items():
    print(f"  {name}: {score}")

# ============================================================
# 任务2：使用 enumerate() 输出序号、姓名和成绩，序号从1开始
# ============================================================
print("\n" + "=" * 60)
print("任务2：enumerate() 带序号输出")
print("=" * 60)

print(f"  {'序号':<6}{'姓名':<10}{'成绩':<6}")
print(f"  {'-'*6}{'-'*10}{'-'*6}")
for idx, name in enumerate(names, start=1):
    print(f"  {idx:<6}{name:<10}{scores[idx-1]:<6}")

# ============================================================
# 任务3：列表推导式——提取及格学生姓名 + 生成成绩平方列表
# ============================================================
print("\n" + "=" * 60)
print("任务3：列表推导式")
print("=" * 60)

# 及格学生姓名（>= 60）
passed_names = [name for name, score in zip(names, scores) if score >= 60]
print(f"及格学生姓名: {passed_names}")

# 所有成绩的平方
squared_scores = [s ** 2 for s in scores]
print(f"成绩平方列表: {squared_scores}")

# ============================================================
# 任务4：计算人数、平均分、最高分、最低分、及格率
# ============================================================
print("\n" + "=" * 60)
print("任务4：统计信息")
print("=" * 60)

num_students = len(names)
avg_score = sum(scores) / num_students
max_score = max(scores)
min_score = min(scores)
pass_rate = sum(1 for s in scores if s >= 60) / num_students * 100

print(f"  人  数: {num_students}")
print(f"  平均分: {avg_score:.2f}")
print(f"  最高分: {max_score}")
print(f"  最低分: {min_score}")
print(f"  及格率: {pass_rate:.1%}")

# ============================================================
# 任务5：比较 alias = scores 与 copied = scores.copy()
# ============================================================
print("=" * 60)
print("任务5：列表赋值（别名）与 copy() 的区别")
print("=" * 60)

# 使用单独的演示列表，避免污染原始成绩数据
demo_scores = [85, 58, 92, 76, 60, 89]

# alias 是同一对象的引用，copied 是浅拷贝产生的新列表
alias = demo_scores
copied = demo_scores.copy()

print(f"  原始列表 demo_scores: {demo_scores}")
print(f"  alias 列表:           {alias}")
print(f"  copied 列表:          {copied}")

# 修改 alias 中的一个元素
alias[0] = 999

print(f"\n  执行 alias[0] = 999 后：")
print(f"  demo_scores: {demo_scores}")   # 会跟着变，因为 alias 和 demo_scores 指向同一对象
print(f"  alias:       {alias}")         # 当然变了
print(f"  copied:      {copied}")        # 不受影响，因为是独立的副本

print("\n  结论：")
print("    alias = demo_scores 只是多了一个名字指向同一个列表对象（引用赋值），")
print("    修改 alias 会直接影响原列表；")
print("    copied = demo_scores.copy() 创建了一个新的列表对象（浅拷贝），")
print("    修改原列表或 alias 都不会影响 copied。")
print(f"    id(demo_scores) == id(alias):   {id(demo_scores) == id(alias)}")
print(f"    id(demo_scores) == id(copied):  {id(demo_scores) == id(copied)}")

# ============================================================
# 任务6：元组不可变性验证 & 不同长度序列 zip() 行为
# ============================================================
print("\n" + "=" * 60)
print("任务6：元组不可变性 & zip() 长度不等时的行为")
print("=" * 60)

# 6.1 元组不可变性
student_record = ("Alice", 85, "A")
print(f"  学生记录元组: {student_record}")
print(f"  尝试修改 student_record[0] = 'Bob' ...")
try:
    student_record[0] = "Bob"
except TypeError as e:
    print(f"    捕获异常: {type(e).__name__}: {e}")
    print("    说明：元组是不可变序列，创建后不能对单个元素重新赋值。")

# 6.2 不同长度序列使用 zip() 的结果
names_short = ["Alice", "Bob", "Carol"]
scores_long = [85, 58, 92, 76, 60, 89]

print(f"\n  姓名列表（3个）: {names_short}")
print(f"  成绩列表（6个）: {scores_long}")

zipped = list(zip(names_short, scores_long))
print(f"\n  zip() 结果: {zipped}")
print(f"  共 {len(zipped)} 对，多余的元素被自动丢弃。")

print("\n  结论：zip() 以最短的序列为准进行配对，较长序列中多出的元素会被忽略。")

print("\n" + "=" * 60)
print("全部任务执行完毕。")
print("=" * 60)
```

实验5：类、继承与对象复制

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score
    
    def is_passed(self):
        """判断是否及格（60分及以上）"""
        return self.score >= 60
    
    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"

    # 有效记录数据
valid_records = [
    ("Alice", 85),
    ("Bob", 58),
    ("Carol", 92),
    ("David", 76),
    ("Eva", 60),
    ("Frank", 89),
]

# 创建Student对象并输出
print("=== 学生信息与及格状态 ===")
students = []
for name, score in valid_records:
    s = Student(name, score)
    students.append(s)
    status = "及格" if s.is_passed() else "不及格"
    print(f"{s}  ->  {status}")

    class GraduateStudent(Student):
        def __init__(self, name, score, research_topic):
            super().__init__(name, score)   # 初始化父类属性
            self.research_topic = research_topic
    
        def __str__(self):
            return (f"GraduateStudent(name={self.name}, score={self.score}, "
                f"research_topic={self.research_topic})")


# 测试
gs = GraduateStudent("Alice", 85, "机器学习")
print(gs)
print(f"是否及格: {gs.is_passed()}")

import copy

def build_original():
    """每次比较都从新建的原始对象开始"""
    return {"name": "Alice", "scores": [85, 90]}

print("=== 直接赋值 ===")
original = build_original()
assigned = original                 # 直接赋值：引用同一对象
assigned["scores"].append(100)      # 修改内部列表
print(f"original: {original}")
print(f"assigned: {assigned}")
print(f"original is assigned: {original is assigned}")
print(f"内部列表同一对象: {original['scores'] is assigned['scores']}")

print("\n=== 浅拷贝 copy.copy() ===")
original = build_original()
shallow = copy.copy(original)       # 浅拷贝：外层字典新建
shallow["scores"].append(100)       # 修改内部列表（仍共享）
print(f"original: {original}")
print(f"shallow : {shallow}")
print(f"original is shallow: {original is shallow}")
print(f"内部列表同一对象: {original['scores'] is shallow['scores']}")

print("\n=== 深拷贝 copy.deepcopy() ===")
original = build_original()
deep = copy.deepcopy(original)      # 深拷贝：递归复制所有层级
deep["scores"].append(100)          # 修改内部列表（独立）
print(f"original: {original}")
print(f"deep    : {deep}")
print(f"original is deep: {original is deep}")
print(f"内部列表同一对象: {original['scores'] is deep['scores']}")
```
