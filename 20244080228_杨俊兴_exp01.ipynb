2.1 编写score_level(score)函数：90〜100分返回“优秀”，80〜不⾜90分返回
“良好”，60〜不⾜80分返回“及格”，0〜不⾜60分返回“不及格”。对于超出0〜
100范围的数值，抛出ValueError

```python
def score_level(score):
    """根据分数返回等级，超出0~100范围抛出ValueError"""
    if not (0 <= score <= 100):
        raise ValueError(f"分数 {score} 超出有效范围 0~100")
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"

# 测试
test_scores = [95, 85, 70, 55, 101, -1]
for s in test_scores:
    try:
        print(f"  分数 {s:>4} -> {score_level(s)}")
    except ValueError as e:
        print(f"  分数 {s:>4} -> 异常: {e}")
```

2.3 编写递归版和迭代版阶乘函数，约定0! = 1，输⼊为⾮负整数。⽐较n = 0、1、
5、10时的结果

```python
def factorial_recursive(n):
    """递归版阶乘，约定 0! = 1"""
    if n < 0 or not isinstance(n, int):
        raise ValueError("输入必须为非负整数")
    if n == 0 or n == 1:
        return 1
    return n * factorial_recursive(n - 1)

def factorial_iterative(n):
    """迭代版阶乘，约定 0! = 1"""
    if n < 0 or not isinstance(n, int):
        raise ValueError("输入必须为非负整数")
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

# 比较 n = 0, 1, 5, 10 时的结果
test_n = [0, 1, 5, 10]
print(f"  {'n':>4} | {'递归版':>12} | {'迭代版':>12} | {'一致':>6}")
print(f"  {'-'*4}-+-{'-'*12}-+-{'-'*12}-+-{'-'*6}")
for n in test_n:
    r_rec = factorial_recursive(n)
    r_itr = factorial_iterative(n)
    match = "✓" if r_rec == r_itr else "✗"
    print(f"  {n:>4} | {r_rec:>12} | {r_itr:>12} | {match:>6}")
```

2.4 编写summary(values)函数，接收⾮空数值列表，以元组返回最⼩值、最⼤值和平
均值，并通过元组解包接收结果

```python
def summary(values):
    """接收非空数值列表，返回 (最小值, 最大值, 平均值) 元组"""
    if not values:
        raise ValueError("列表不能为空")
    min_val = min(values)
    max_val = max(values)
    avg_val = sum(values) / len(values)
    return (min_val, max_val, avg_val)

# 测试并通过元组解包接收结果
data = [12, 45, 7, 23, 89, 34, 56]
min_v, max_v, avg_v = summary(data)
print(f"  数据: {data}")
print(f"  最小值: {min_v}")
print(f"  最大值: {max_v}")
print(f"  平均值: {avg_v:.2f}")
```

实验5：类、继承与对象复制

5.1 定义Student类，在__init__()中初始化name和score属性，并实现is_passed()
和__str__()⽅法

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score
    
    def is_passed(self):
        """判断是否及格（60分及以上）"""
        return self.score >= 60
    
    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"
```

5.2 为实验4中的每条有效记录创建⼀个Student对象，输出对象信息和及格状态

```python
# 实验4中的有效记录数据
valid_records = [
    ("Alice", 85),
    ("Bob", 58),
    ("Carol", 92),
    ("David", 76),
    ("Eva", 60),
    ("Frank", 89),
]

print("=== 学生信息与及格状态 ===")
students = []
for name, score in valid_records:
    s = Student(name, score)
    students.append(s)
    status = "及格" if s.is_passed() else "不及格"
    print(f"{s}  ->  {status}")
```

5.3 定义继承⾃Student的GraduateStudent类，增加research_topic属性，通过
super().__init__()初始化⽗类属性，并重写__str__()

```python
class GraduateStudent(Student):
    """研究生类：继承自 Student，增加 research_topic 属性"""

    def __init__(self, name, score, research_topic):
        super().__init__(name, score)        # 初始化父类属性
        self.research_topic = research_topic  # 新增属性

    def __str__(self):
        """重写 __str__()，加入研究方向"""
        return (f"GraduateStudent(name={self.name}, score={self.score}, "
                f"research_topic={self.research_topic})")


# 测试
gs = GraduateStudent("Alice", 85, "机器学习")
print(gs)
print(f"是否及格: {gs.is_passed()}")
```

5.4 构造包含嵌套列表的对象，例如{"name": "Alice", "scores": [85, 90]}。分
别进⾏直接赋值、copy.copy()和copy.deepcopy()，修改内部列表，⽐较结果。
每次⽐较都从新建的原始对象开始

```python
import copy


def build_original():
    """每次比较都从新建的原始对象开始"""
    return {"name": "Alice", "scores": [85, 90]}


# ---------- 4.1 直接赋值 ----------
print("=== 直接赋值 ===")
original = build_original()
assigned = original                 # 直接赋值：引用同一对象
assigned["scores"].append(100)      # 修改内部列表
print(f"original: {original}")
print(f"assigned: {assigned}")
print(f"original is assigned: {original is assigned}")
print(f"内部列表同一对象: {original['scores'] is assigned['scores']}")


# ---------- 4.2 浅拷贝 copy.copy() ----------
print("\n=== 浅拷贝 copy.copy() ===")
original = build_original()
shallow = copy.copy(original)       # 浅拷贝：外层字典新建
shallow["scores"].append(100)       # 修改内部列表（仍共享）
print(f"original: {original}")
print(f"shallow : {shallow}")
print(f"original is shallow: {original is shallow}")
print(f"内部列表同一对象: {original['scores'] is shallow['scores']}")


# ---------- 4.3 深拷贝 copy.deepcopy() ----------
print("\n=== 深拷贝 copy.deepcopy() ===")
original = build_original()
deep = copy.deepcopy(original)      # 深拷贝：递归复制所有层级
deep["scores"].append(100)          # 修改内部列表（独立）
print(f"original: {original}")
print(f"deep    : {deep}")
print(f"original is deep: {original is deep}")
print(f"内部列表同一对象: {original['scores'] is deep['scores']}")
```
